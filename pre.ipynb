Проводим EDA, добавляем новые колонки для удобства построения графиков

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='mixed', dayfirst=True)
df['Ship Date']  = pd.to_datetime(df['Ship Date'],  format='mixed', dayfirst=True)

df['Shipping Duration'] = (df['Ship Date'] - df['Order Date']).dt.days

df = df[df['Shipping Duration'] >= 0]

df_clean = df[['Order ID', 'Order Date', 'Category', 'Sub-Category', 'City', 'Country', 
               'Sales', 'Profit', 'Shipping Duration', 'Quantity']]


df_clean['Year'] = df['Order Date'].dt.year
df_clean['Month'] = df['Order Date'].dt.month_name()


df_clean['Profit Margin'] = (df_clean['Profit'] / df_clean['Sales']) * 100

df_clean.to_csv('dashboard_ready.csv', index=False)
```
